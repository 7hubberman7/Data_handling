# Criação do DataFrame de Livros

```python
import pandas as pd

# Criando um DataFrame com dados de vendas de livros
book_data = {
    "book_id": [201, 202, 203, 204, 205],
    "title": ["Python for Beginners", "Data Science with Python", "Advanced Python Programming", "Learning SQL", "Python for Data Analysis"],
    "author": ["John Doe", "Jane Smith", "John Doe", "Alice Johnson", "Jane Smith"],
    "price": [250, 300, 150, 180, 280],
    "copies_sold": [150, 200, 100, 90, 300]
}

books_df = pd.DataFrame(book_data)
print(books_df)
```

## Filtrando Dados por Preço

```python
price_200_df = books_df[books_df["price"] > 200]
print(price_200_df)
```

## Filtrando por Múltiplas Condições

```python
valor_menor_200_df = books_df[(books_df["price"] <= 200) & (books_df["copies_sold"] >= 100)]

print(valor_menor_200_df)
```

## Filtrando por Autor

```python
authors_df = books_df[books_df["author"].isin(["John Doe", "Jane Smith"])]
print(authors_df)
```

## Filtrando por Título

```python
titles = books_df[books_df["title"].str.contains("Python")]
print(titles)
```
