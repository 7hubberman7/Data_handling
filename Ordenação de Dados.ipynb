```python
import pandas as pd

# Criando um DataFrame com dados de funcionários
employee_data = {
    "employee_id": [101, 102, 103, 104, 105],
    "name": ["Alice", "Bob", "Charlie", "David", "Eve"],
    "department": ["HR", "IT", "Finance", "Marketing", "IT"],
    "salary": [50000, 60000, 55000, 70000, 65000],
    "years_of_experience": [5, 7, 4, 10, 6]
}

employees_df = pd.DataFrame(employee_data)
print(employees_df)
```

# Ordenado por salario em ordem crescente(padrão)

```python
order_salario = employees_df.sort_values(by="salary")
print(order_salario)
```

# Ordenando por experiencia em ordem decrescente

```python
sort_by_ex = employees_df.sort_values(by="years_of_experience", ascending=False)
print(sort_by_ex)
```

# Ordenando por departamento em ordem decrescente e salario em ordem crescente

```python
depart = employees_df.sort_values(by=["department","salary"], ascending=[False,True])
print(depart)
```
