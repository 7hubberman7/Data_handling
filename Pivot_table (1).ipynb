```markdown
Este notebook demonstra a criação e manipulação de um DataFrame Pandas para analisar dados de vendas de produtos. Ele inclui exemplos de como calcular totais de vendas por categoria, vendas por produto e região, e a média de vendas.

### Conteúdo:
- Criação de um DataFrame de exemplo com dados de vendas.
- Análise de vendas por categoria.
- Análise de vendas por produto e região.
- Cálculo do total e da média de vendas por produto.
```

````python
import pandas as pd

# Criando um conjunto de dados
data = {
    "order_id": [201, 202, 203, 204, 205, 206, 207, 208, 209, 210],
    "product": [
        "Laptop",
        "Smartphone",
        "Laptop",
        "Tablet",
        "Smartphone",
        "Tablet",
        "Smartwatch",
        "Laptop",
        "Headphones",
        "Smartwatch",
    ],
    "category": [
        "Electronics",
        "Electronics",
        "Electronics",
        "Electronics",
        "Electronics",
        "Electronics",
        "Wearables",
        "Electronics",
        "Accessories",
        "Wearables",
    ],
    "region": [
        "North",
        "South",
        "East",
        "North",
        "West",
        "South",
        "East",
        "West",
        "North",
        "West",
    ],
    "sales": [1500, 900, 1200, 600, 750, 450, 320, 1400, 200, 380],
}

df = pd.DataFrame(data)
print(df)
````

# Criando tabela dinâmica

````python
tabela_dinamica = pd.pivot_table(df, values='sales', index="category", aggfunc='sum').sort_values(by= "sales", ascending=False)
print(tabela_dinamica)
````

A categoria que mais vende é a de eletronicos com 6800

# total de vendas para cada produto por região


````python
ven_reg = pd.pivot_table(df, values='sales', index="product", columns="region", aggfunc='sum').sort_values(by= "North", ascending=False)
print(ven_reg)
````

Na região norte o produto que apresenta maior valor em vendas são os laptops com 1500.

# total de vendas e a média de vendas para cada categoria de produto

````python
total_vendas = pd.pivot_table(df, values='sales', index="product", aggfunc=['sum', "mean"]).sort_values(by=("sum", "sales"), ascending=False)
print(total_vendas)
````

````python
vendas = pd.pivot_table(df, values='sales', index="product", columns="region", aggfunc='sum', fill_value=0)
print(vendas)
````

 dinâmica utilizando o parâmetro fill_value=0 que mostre o total de vendas para cada produto e região.
