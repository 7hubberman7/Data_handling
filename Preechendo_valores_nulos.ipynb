```python
import pandas as pd

# Criando um DataFrame
data = {
    "order_id": [1001, 1002, 1003, 1004, 1005, 1006, 1007],
    "customer_name": ["Alice", "Bob", None, "David", "Eva", "Frank", None],
    "product_category": ["Electronics", "Clothing", "Electronics", "Books", None, "Clothing", "Books"],
    "quantity": [2, 1, None, 1, 3, None, 4],
    "unit_price": [299.99, 49.99, 199.99, None, 15.99, 79.99, 12.99],
    "order_date": ["2024-08-01", "2024-08-02", "2024-08-03", "2024-08-04", "2024-08-05", "2024-08-06", None]
}

df = pd.DataFrame(data)
print(df)
```

```python
df.isna().sum()
```

### Calcular porcentagem de valores ausentes por linha e coluna

```python
missing_col_percentage = df.isnull().sum() / len(df) * 100
print("Porcentagem de valores ausentes por coluna:")
print(missing_col_percentage)

missing_row_percentage = df.isnull().sum(axis=1) / len(df.columns) * 100
print("\nPorcentagem de valores ausentes por linha:")
print(missing_row_percentage)
```

### Remover linhas/colunas com mais de 50% de valores ausentes

Neste conjunto de dados, nenhuma linha ou coluna excede 50% de valores ausentes, portanto, nenhuma remoção será feita. O código abaixo demonstra como isso seria feito se houvesse.

```python
# Identificar colunas com mais de 50% de valores ausentes
threshold_col = 50
cols_to_drop = missing_col_percentage[missing_col_percentage > threshold_col].index

# Identificar linhas com mais de 50% de valores ausentes
threshold_row = 50
rows_to_drop = missing_row_percentage[missing_row_percentage > threshold_row].index

# Remover colunas e linhas, se houver
if not cols_to_drop.empty:
    df.drop(columns=cols_to_drop, inplace=True)
    print(f"Colunas removidas: {list(cols_to_drop)}")
else:
    print("Nenhuma coluna com mais de 50% de valores ausentes para remover.")

if not rows_to_drop.empty:
    df.drop(index=rows_to_drop, inplace=True)
    print(f"Linhas removidas: {list(rows_to_drop.index)}")
else:
    print("Nenhuma linha com mais de 50% de valores ausentes para remover.")
```

### Preencher valores ausentes em colunas específicas

```python
# 1. Preencher 'customer_name' com 'Unknown'
df['customer_name'].fillna('Unknown', inplace=True)

# 2. Preencher 'product_category' com 'Miscellaneous'
df['product_category'].fillna('Miscellaneous', inplace=True)

# 3. Preencher 'quantity' e 'unit_price' com a média
df['quantity'].fillna(df['quantity'].mean(), inplace=True)
df['unit_price'].fillna(df['unit_price'].mean(), inplace=True)

# 4. Preencher 'order_date' com a última data de pedido conhecida
# Primeiro, converta a coluna para datetime
df['order_date'] = pd.to_datetime(df['order_date'])

# Encontre a última data de pedido conhecida (a data máxima)
last_known_date = df['order_date'].max()

# Preencha os valores ausentes com a última data conhecida
df['order_date'].fillna(last_known_date, inplace=True)

print("DataFrame após o tratamento de valores ausentes:")
display(df)
```
