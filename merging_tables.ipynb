# Demonstração de Operações JOIN com Pandas

Este notebook demonstra as diferentes operações de `JOIN` (`INNER`, `LEFT`, `OUTER`) usando a biblioteca Pandas em Python. Utiliza dois DataFrames (`df_products` e `df_order_details`) para ilustrar como combinar informações de diferentes tabelas com base em uma chave comum.

```python
import pandas as pd

# Criando um DataFrame com dados dos produtos
products = {
    "product_id": [1, 2, 3, 4],
    "product_name": ["Laptop", "Tablet", "Smartphone", "Smartwatch"],
    "category": ["Electronics", "Electronics", "Electronics", "Wearables"],
    "price": [1200, 300, 800, 250],
}

df_products = pd.DataFrame(products)
print(df_products)

# Criando um DataFrame com os detalhes dos pedidos
order_details = {
    "order_id": [101, 102, 103, 104, 105, 106, 107],
    "product_id": [2, 2, 1, 2, 4, 1, 5],
    "customer_id": [201, 202, 202, 204, 203, 201, 202],
    "quantity": [1, 2, 1, 3, 2, 2, 4],
}

df_order_details = pd.DataFrame(order_details)
print(df_order_details)
```

### Preparação dos Dados

Primeiro, vamos criar dois DataFrames: um para `produtos` (`df_products`) e outro para `detalhes de pedidos` (`df_order_details`). Estes DataFrames serão usados para demonstrar as operações de `JOIN`.

# INNER JOIN entre a tabela de pedidos e a tabela de produtos

### INNER JOIN

O `INNER JOIN` retorna apenas as linhas que possuem correspondência (valores iguais na coluna `product_id`) em ambos os DataFrames. Linhas que não encontram correspondência em nenhum dos DataFrames são descartadas.

Neste exemplo, veremos apenas os pedidos que correspondem a produtos existentes na tabela `df_products`.

```python
inner = pd.merge(df_order_details, df_products, on='product_id', how='inner')
print(inner)
```

# LEFT JOIN entre a tabela de pedidos e a tabela de produtos

### LEFT JOIN

O `LEFT JOIN` (ou `LEFT OUTER JOIN`) retorna todas as linhas do DataFrame esquerdo (`df_order_details` neste caso) e as linhas correspondentes do DataFrame direito (`df_products`). Se não houver correspondência no DataFrame direito, os valores das colunas do DataFrame direito serão preenchidos com `NaN` (Not a Number).

Podemos observar que o `product_id` 5 do `df_order_details` não possui um produto correspondente em `df_products`, resultando em `NaN` para as colunas de produto.

```python
left = pd.merge(df_order_details, df_products, on='product_id', how='left')
print(left)
```

não há registro de vendas de smartphones

# outer join da tabela de produtos e pedidos


### OUTER JOIN

O `OUTER JOIN` (ou `FULL OUTER JOIN`) retorna todas as linhas quando há uma correspondência em um dos DataFrames. Ou seja, ele inclui todas as linhas de ambos os DataFrames, preenchendo com `NaN` onde não há correspondência.

Neste exemplo, ele incluirá:
- Pedidos que têm produtos correspondentes.
- Produtos que não têm pedidos correspondentes (ex: `product_id` 3 - Smartphone).
- Pedidos que não têm produtos correspondentes (ex: `product_id` 5).

```python
outer = pd.merge(df_order_details, df_products, on='product_id', how='outer')
print(outer)
```

os produto de id "5" está ausente na tabela de produto
